```python
import numpy as np
import pandas as pd
```

```python
data = [10, 20, 30, 40, 50, 60]

mean = np.mean(data)
median = np.median(data)
variance = np.var(data)

mean, median, variance
```

```python
# Probability of getting even number from list
numbers = [1, 2, 3, 4, 5, 6]

even = [n for n in numbers if n % 2 == 0]

probability_even = len(even) / len(numbers)
probability_even
```

```python
data = {
    "Hours_Studied": [2, 4, 6, 8, 10],
    "Marks": [40, 50, 65, 75, 90]
}

df = pd.DataFrame(data)

df.corr()
```

```python
vector_a = np.array([1, 2, 3])
vector_b = np.array([4, 5, 6])

vector_sum = vector_a + vector_b
dot_product = np.dot(vector_a, vector_b)

vector_sum, dot_product
```

```python
matrix_a = np.array([[1, 2], [3, 4]])
matrix_b = np.array([[5, 6], [7, 8]])

matrix_add = matrix_a + matrix_b
matrix_mul = np.dot(matrix_a, matrix_b)

matrix_add, matrix_mul
```

### Statistics & Math for ML – Concept Notes

- Mean: Average value
- Median: Middle value
- Variance: Measure of spread
- Probability: Likelihood of an event
- Correlation: Relationship between variables
- Linear Algebra: Foundation of ML models
